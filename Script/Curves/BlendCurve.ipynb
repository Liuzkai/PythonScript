```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

# 1. Linear
def alpha_linear(t):
    return t

plt.figure(figsize=(10, 10))
plt.plot(t, alpha_linear(t), 'r-', label='Linear')
plt.title('1. Linear Interpolation')
plt.xlabel('Time (t)')
plt.ylabel('Alpha(t)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 2. Cubic-in
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

def alpha_cubic_in(t):
    return t ** 3

plt.figure(figsize=(10, 6))
plt.plot(t, alpha_cubic_in(t), 'r-', label='Cubic-in')
plt.title('2. Cubic-in Interpolation')
plt.xlabel('Time (t)')
plt.ylabel('Alpha(t)')
plt.grid(True)
plt.legend()
plt.show()

```

```python
# 3. HermiteCubic
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

def alpha_hermite_cubic(t):
    return t**2 * (3 - 2*t)

plt.figure(figsize=(10, 6))
plt.plot(t, alpha_hermite_cubic(t), 'r-', label='HermiteCubic')
plt.title('3. HermiteCubic Interpolation')
plt.xlabel('Time (t)')
plt.ylabel('Alpha(t)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 4. Sinusoidal
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

def alpha_sinusoidal(t):
    return 1 - np.cos(np.pi * t / 2)

plt.figure(figsize=(10, 6))
plt.plot(t, alpha_sinusoidal(t), 'r-', label='Sinusoidal')
plt.title('4. Sinusoidal Interpolation')
plt.xlabel('Time (t)')
plt.ylabel('Alpha(t)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 5. QuadraticInOut
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

def alpha_quadratic_in_out(t):
    return np.where(t < 0.5,
                    2 * t * t,
                    -2 * t * t + 4 * t - 1 )

plt.figure(figsize=(10, 6))
plt.plot(t, alpha_quadratic_in_out(t), 'r-', label='QuadraticInOut')

plt.title('5. QuadraticInOut Interpolation')
plt.xlabel('Time (t)')
plt.ylabel('Alpha(t)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate t values
t = np.linspace(0, 1, 100)

# Different k values to plot
k_values = [1, 2, 5, 10]

def transition_curve(t, k):
    """
    Computes the smooth transition curve from 0 to 1.

    Parameters:
    t (float or array-like): The input value(s) between 0 and 1.
    k (float): The parameter controlling the smoothness of the transition.

    Returns:
    float or array-like: The output value(s) of the transition curve.

    Example:
    >>> transition_curve(0.5, 2)
    0.5
    """
    return t**k / (t**k + (1 - t)**k)


# Plot the curves
for k in k_values:
    y = transition_curve(t, k)
    plt.plot(t, y, label=f'k={k}')

# Add legend and show plot
plt.legend()
plt.show()


```
